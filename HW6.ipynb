### W6A16
Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
#nhập n, m
n, m = map(int, input().split())
matrix = []
for i in range (n):
    row = list(map(int, input().split())) #nhập danh sách là các cột ở hàng i -> hàng i
    matrix.append(row) # thêm hàng i vào matrix
for row in matrix:
    for item in row:
        print(f"{item:>4}", end = "") #> căn lề phải
        # 4: độ rộng là 1 tab
    print() #in dòng trắng để xuống dòng sau khi in 1 hàng
```

### W6A17
Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input())
#tạo ma trận
matrix = []
for i in range (n):
    row = list(map(int, input().split()))
    matrix.append(row)
# tìm đường chéo chính phụ
main = []
side = []
for i in range(n):
    main.append(matrix[i][i])
    side.append(matrix[i][n -1 - i])
print(main)
print(side) 
```

### A19
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def vt_max(lst):
    max_val = lst[0]
    max_idx = 0
    for i in range (1, len(lst)):
        if lst[i] > max_val:
            max_val = lst[i]
            max_idx = i
    return max_idx

ds = list(map(int, input().split()))
vi_tri = vt_max(ds)
print(ds)
print("Vị trí của phần tử có giá trị lớn nhất trong list:", vi_tri)
```

### A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def vt_k(lst, k):
    for i in range (len(lst)):
        if lst[i] == k:
            return i
    return -1 # trả về -1 nếu chạy hết vòng lặp mà không thấy k
ds = list(map(int, input().split()))
k = int(input())
print(vt_k(ds, k))
```

```python
def vt_k(lst, k):
    if k in lst:
        return lst.index(k) #phương thức.index(): lấy vị trí của phần tử k trong lst
    return -1
```

### A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
cac_chuoi = input().split()
d ={}
for index, chuoi in enumerate(cac_chuoi):
    if chuoi not in d:
        d[chuoi] = [index]
    else:
        d[chuoi].append(index)
# chuyển danh sách vị trí thành tuple
for key in d:
    d[key] = tuple(d[key])
print(d)
```

```python
def so_lon_hon_x(arr, x):
    ket_qua = [so for so in arr if so >= x] 
    return ket_qua
mang = [3, 1, 5, 7, 4]
x = 4
print(so_lon_hon_x(mang, x))
```

```python
def xoa_so_nho(arr, x):
    for i in range (len(arr)-1, -1, -1): # duyệt ngược sẽ tránh thay đổi chỉ số
        # khi xóa trực tiếp trên mảng
        if arr[i] < x:
            arr.pop(i) # xóa index thứ i
    return arr
```

### A24
Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9


```python
def day_so(lst):
    khong = [x for x in lst if x == 0]
    khac = [x for x in lst if x != 0]
    return khong + khac
lst = list(map(int, input().split()))
print(day_so(lst))
```

### A26

phần tử đầu và cuối

### A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
arr = input().split()
tang =True
giam = True
for i in range (1, len(arr)):
    if arr[i] >= arr[i-1]:
        giam = False
    if arr[i] <= arr[i-1]:
        tang = False
if tang:
    print("Dãy tăng thực sự")
elif giam:
    print("Dãy giảm thực sự")
else:
    print("Dãy vô danh")
```
