### A9
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới
trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có
trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của
key

```python
#eval(expression)
# hàm thực hiện yêu cầu của expression
#nhập dict theo định dạng dict
d1 = eval(input()) #nhập, input dạng dict -> eval thêm dict vào, d1 có ype là dict
d2 = eval(input())
dgop = d1.copy() #tạo bản sao mà khi thao tác không ảnh hưởng đến dict gốc
for key, value in d2.items(): #.items(): duyệt key, value của dict theo từng cặp
    if key in dgop:
        dgop[key] += value # dgop[key]: lấy value của key ở list dgop
    else:
        dgop[key] = value
dket_qua = dict(sorted(dgop.items()))
print(dket_qua)
```

### A10
Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra
list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple
có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của
số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên
trong tuple

```python
l = list(map(int, input().split()))
#khởi tạo danh sách các số nguyên
#nhập các số nguyên trên 1 dòng, sau đó ép kiểu thành list
lt = [] #tạo list rỗng để sau đó thêm tuple vào
k = int(input())
n =len(l) #tạo range để lát chạy vòng lặp for thì chỉ phụ thuộc 1 biến chạy, dễ kiểm soát biến i đúng trước i + 1
cap_da_co = [] #tạo danh sách rỗng để lát kiểm tra tuple có lặp lại hay không
for i in range (n):
    for j in range (i+1, n):
        if (l[i] + l[j] == k):
            cap = (l[i], l[j])
            if cap not in cap_da_co:
                lt.append(cap)# thêm tuple cap vào danh sách lt
                cap_da_co.append(cap)
print(lt)
```

### A12
[Tìm phần tử xuất hiện nhiều nhất]
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list.
Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất
trong số đó

```python
l =list(map(int, input().split())) #nhập dữ liệu có dạng list
d ={} #tạo dict rỗng để lưu số lần xuất hiện của từng số trong list
for i in l:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1
#tìm số xuất hiện nhiều nhất
max_cnt = 0
for so, so_lan in d.items():
    if so_lan > max_cnt:
        max_cnt = so_lan
#tìm số nhỏ nhất trong các số có cùng số lần xuất hiện nhiều nhất
ket_qua = None
for so, so_lan in d.items(): #duyệt từng cặp key-value trong d
    if so_lan == max_cnt: 
        if (ket_qua is None) or (so < ket_qua):
            ket_qua = so
print(ket_qua)      
```

```python
lst = list(map(int, input().split()))
d = {}
max_val = lst[0]
max_count = 1
for i in lst:
    if i not in d:
        d[i] = 1
    else:
        d[i] += 1
    # so sánh giá trị của các số có số lần xuất hiện nhiều nhất giionsg nhau 
    if d[i] > max_count:
        max_count = d[i]
        max_val = i
    elif d[i] == max_count and i < max_val:
        max_val = i
print(d)
print(max_val)
```

### A13
[Đảo ngược key và value của dictionary]
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một
dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In
ra dictionary mới.

```python
d = input().split()
#tạo dict từ danh sách đã tách
dict_goc = {} # tạo dict rỗng
for i in range (0, len(d), 2): #chạy trong range chiều dài của d lấy giá trị của index i,
    #bước nhảy là 2 sẽ lấy được key là giá trị của phần tử index i
    key = d[i]
    value = d[i+1]
    dict_goc[key] = value #khởi tạo key tên là key với vue là value, lưu key-value vào dict
dict_dao = {v: k for k,v in dict_goc.items()}
print(d)
print(dict_dao)
```

### A15
[Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số
nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp keyvalue từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới

```python
lst = input().split()
d = {}
for i in range (0, len(lst), 2):
    key = lst[i]
    value = int(lst[i+1])
    d[key] = value
k = int(input())
d_moi = {key: value for key, value in d.items() if value > k}
print(d)
```
